```python
import pandas as pd
```

```python
df1 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv")
df1.head()
```

```python
df2 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file2.csv")
df2.head()
```

```python
df3= pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file3.csv")
df3.head()
```

```python
df1.columns = df1.columns.str.lower().str.strip()
df2.columns = df2.columns.str.lower().str.strip()
df3.columns = df3.columns.str.lower().str.strip()


df1 = df1.rename(columns={'st': 'state', 'gender': 'gender'}) 
df2 = df2.rename(columns={'st': 'state', 'gender': 'gender'})
df3 = df3.rename(columns={'st': 'state', 'gender': 'gender'})

```

```python
df = pd.concat([df1, df2, df3], axis=0, ignore_index=True)
```

```python
df.info()
```

```python
df.rename(columns={col : col.replace(" ", "_").lower() for col in df.columns}, inplace = True)
```

```python
gender_map = {
    'f': 'F', 'female': 'F', 'femal': 'F',
    'm': 'M', 'male': 'M',
}

df['gender'] = df['gender'].replace(gender_map)
```

```python
state_rep = {
    "AZ": "Arizona", "Cali": "California", "WA": "Washington"
}

df['state'] = df['state'].replace(state_rep)
df.head(10)
```

```python
df["customer_lifetime_value"] = df["customer_lifetime_value"].apply(lambda x: float(str(x).replace('%', '')) if pd.notnull(x) else x)
```

```python
complaints_ = {
    "1/0/00": "0",
    "1/2/00": "2",
    "1/1/00": "1",
    "1/3/00": "3",
    "1/4/00": "4",
    "1/5/00": "5"}

df['number_of_open_complaints'] = df['number_of_open_complaints'].replace(complaints_)
```

```python
df["number_of_open_complaints"] = pd.to_numeric(df["number_of_open_complaints"], errors='coerce')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna(how='all', inplace=True)
```

```python
df.isnull().sum()
```

```python
df['gender'] = df['gender'].fillna('unknown')
df.head()
```

```python
# Your code goes here

df4 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv")
df4.head()
```

1. You work at the marketing department and you want to know which sales channel brought the most sales in terms of total revenue. Using pivot, create a summary table showing the total revenue for each sales channel (branch, call center, web, and mail).
Round the total revenue to 2 decimal points.  Analyze the resulting table to draw insights.

```python
print(f"  → Rows: {df4.shape[0]:,}  |  Columns: {df4.shape[1]}")
print()
```

```python
print(f"  → unique channels in 'sales_channel': {df4['sales_channel'].unique().tolist()}")
print(f"  → type of 'total_claim_amount': {df4['total_claim_amount'].dtype}")
print(f"  → null values in 'total_claim_amount': {df4['total_claim_amount'].isnull().sum()}")
print()
```

```python
pivot = pd.pivot_table(
    data=df4,
    values="total_claim_amount",  
    index="sales_channel",         
    aggfunc="sum",                 
)
```

```python
pivot = pivot.round(2)
pivot.columns = ["Total Revenue"]
```

```python
pivot = pivot.sort_values("Total Revenue", ascending=False)
print(pivot)
```

```python
total_global = pivot["Total Revenue"].sum()
pivot["% from total"] = ((pivot["Total Revenue"] / total_global) * 100).round(2)
```

```python
print(pivot.to_string())
print()
```

2. Create a pivot table that shows the average customer lifetime value per gender and education level. Analyze the resulting table to draw insights.

```python
pivot2 = pd.pivot_table(
    data=df4,
    values="customer_lifetime_value",
    index="education",            
    columns="gender",             
    aggfunc="mean",               
)
```

```python
pivot2 = pivot2.round(2)
 
print(pivot2.to_string())
print()
```

```python
pivot2["Mean"] = pivot2.mean(axis=1).round(2)
pivot2 = pivot2.sort_values("Mean", ascending=False)
print(pivot2.to_string())
print()
```

*In data analysis, a long format table is a way of structuring data in which each observation or measurement is stored in a separate row of the table. The key characteristic of a long format table is that each column represents a single variable, and each row represents a single observation of that variable.*

*More information about long and wide format tables here: https://www.statology.org/long-vs-wide-data/*
